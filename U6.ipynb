#### write a program to find sum of n numbers. n is to be enter by user

```python
n=int(input("Enter a number:"))
sum=0
for i in range(1,n+1):
    sum=sum+i
print("sum =",sum)
```

#### write a program to read 3 numbers from user and check how many numbers available between A and B are divisible by C.

```python
A=int(input("Enter a A:"))
B=int(input("Enter a B:"))
C=int(input("Enter a C:"))
count=0
for i in range(A+1,B):
    if i%C==0:
        count=count+1
print(count)
```

#### write a program to print fibonaci sequence up to n term. n is entered by user and n is number of term

```python
n=int(input("Enter a number"))
n1=0
n2=1
if n==1:
        print(n1,n2)
else:
        print("Fibonaci sequence")
        print(n1,end=" ")
        print(n2,end=" ")
        for i in range(n-2):
            nth=n1+n2
            print(nth,end=" ")
            n1=n2
            n2=nth
```

#### write a program to take 10 numbers from user and print them sum and avg.

#### write a program to find reverse number from user entered by given.

```python
n=int(input("Enter a number:"))
rev_n=0
while n!=0:
    digit=n%10
    rev_n=rev_n*10+digit
    n=n//10
print("Reverse number =",rev_n)
```
