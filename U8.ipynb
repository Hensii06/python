### Variable length Argument

```python
def fun(n1,*s):
    print(n1)
    for s1 in s:
        print(s1)
fun(10)
fun(10,20,30)
fun(10,"A","B",20,30)
```

```python
def fun(*s,n1):
    for s1 in s:
        print(s1)
    print(n1)
fun(10)
fun(10,20,30)
fun(10,"A","B",20,30)
```

```python
def fun(*s,n1):
    for s1 in s:
        print(s1)
    print(n1)
fun(10,20,n1=30)
fun(10,"A","B",20,n1=30)
```

#### *n -> args argument   data type -> tuple
#### **n -> keyword argument    data type -> dix.

```python
def fun(**n):
    print(n)
    for k,v in n.items():
        print(k,"=",v)
fun(m1=100,m2=200,m3=300,m4=400)
fun(rno=1,name='urpit',marks=80)
```

### nested function

```python
def outer():
    print("this is outer function")
    def inner():
        print("this is inner function")
    return inner()
outer()
```

```python
def happy(n):
    seen=set()
    while n!=1 and n not in seen:
        seen.add(n)
        total_sum=0
        while n>0:
            total_sum+=(n%10)**2
            n//=10
        n=total_sum
    if n==1:
        print("number is happy")
    else:
        print("number is not happy")
n=int(input("number"))
happy(n)
```

### Harshad number

```python
def harshad(n):
    sum=0
    num=n
    while n>0:
        sum=sum+(n%10)
        n=n//10
    if num%sum==0:
        print("Number is harshad number")
    else:
        print("Number is not harshad number ")
n=int(input("number"))
harshad(n)
```
