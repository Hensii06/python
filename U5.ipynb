### while loop Syntax
while condition:

    body of loop
    increment/decrement

```python
x=1
while x<=5:
    print(x)
    x+=1
```

```python
for i in range(2):
    for j in range(2):
        print(i,j)
```

```python
name=""
while name!="urpit":
    name=input("Enter name")
print("thanks for confirmation")
```

```python
n=[1,2,3,4,5]
for i in n:
    print(i)
else:
    print("Loop completed without finding a specific condition")
```

```python
n=[1,2,3,4,5]
for i in n:
    if i==3:
        print("Found 3 ,break the loop")
        break
    print(i)
else:
    print("Loop completed without finding a specific condition")
```

#### write a program to check if a given year is leap year or not 

```python
A=int(input("enter year"))
if (A%4==0 and A%100!=0) or (A%400==0):
    print(A,"is leap year")
else:
    print(A,"is not leap year")
```
