### write a program to get next day of a given day

```python
year=int(input("Enter year"))
leap_year=False
if year%4==0 and (year%400==0 or year%100!=0):
    leap_year=True
month=int(input("Enter month"))
if month in (1,3,5,7,8,10,12):
    length=31
elif month==2:
    if leap_year:
        length=29
    else:
        length=28
else:
    length=30
day=int(input("Enter day"))
if day>length:
    print("Enter valid day")
else:
    if day<length:
        day=day+1
        print(f"next date:{year}-{month}-{day}")
    else:
        day=1
        if month==12:
            month=1 
            year=year+1
        else:
            month=month+1
        print(f"next date:{year}-{month}-{day}")
```
