## Global v/s Local Variable

```python
a5="Python"
def demo():
    global a5
    a5="Java"
    print(a5)
demo()
print(a5)
```

```python
a5="Python"
def demo():
    a5="Java"
    print(a5)
demo()
print(a5)
```
